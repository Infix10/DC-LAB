```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(-5,5,0.01)
unit_step=(t>=0).astype(float)
unit_impulse = (t==0).astype(float)
ramp = t*(t>=0).astype(float)
sine_wave = np.sin(2*np.pi*1*t)
exponential = np.exp(-0.5*t)*(t>=0).astype(float)
fig,axs = plt.subplots(5,1,figsize=(10,10))
#Unit step
axs[0].plot(t,unit_step,linewidth=2,color='b')
axs[0].set_title('Unit Step Signal')
axs[0].grid(True)
#Unit impulse
axs[1].stem(t,unit_impulse,linefmt='r',markerfmt='ro',basefmt='')
axs[1].set_title('Unit Impulse Signal')
axs[1].grid(True)
#Ramp
axs[2].plot(t,ramp,linewidth=2,color='g')
axs[2].set_title('Ramp Signal')
axs[2].grid(True)
#Sine wave
axs[3].plot(t,sine_wave,linewidth=2,color='r')
axs[3].set_title('Sine Wave')
axs[3].grid(True)
#Exponential
axs[4].plot(t,exponential,linewidth=2,color='y')
axs[4].set_title('Exponential Decay Signal')
axs[4].grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Time axis (continuous approximation)
t = np.linspace(0, 2*np.pi, 1000)

# Define two continuous-time signals
x1 = np.sin(t)
x2 = np.cos(t)

# Add the signals
y = x1 + x2

# Plot
plt.figure(figsize=(10, 5))
plt.plot(t, x1, label='x1(t) = sin(t)')
plt.plot(t, x2, label='x2(t) = cos(t)')
plt.plot(t, y, label='y(t) = x1(t) + x2(t)', linewidth=2)

plt.title('Continuous-Time Signal Addition')
plt.xlabel('Time (t)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Discrete-time index
n = np.arange(0, 20)

# Define two discrete-time signals
x1 = np.sin(0.3 * np.pi * n)
x2 = np.cos(0.3 * np.pi * n)

# Add the signals
y = x1 + x2

# Plot
plt.figure(figsize=(10, 5))

plt.stem(n, x1, linefmt='b-', markerfmt='bo', basefmt='k-', label='x1[n]')
plt.stem(n, x2, linefmt='g-', markerfmt='go', basefmt='k-', label='x2[n]')
plt.stem(n, y, linefmt='r-', markerfmt='ro', basefmt='k-', label='y[n] = x1[n] + x2[n]')

plt.title('Discrete-Time Signal Addition')
plt.xlabel('n')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.show()
```
